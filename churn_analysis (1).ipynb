```python
import matplotlib.pyplot as plt
import seaborn as sns
counts = pd.crosstab(df['SeniorCitizen'], df['Churn'])

pct_to_total = (counts / counts.sum().sum()) * 100

ax = pct_to_total.plot(kind='bar', stacked=True, figsize=(4,4), color=['skyblue', 'salmon'])

for p in ax.patches:
    width, height = p.get_width(), p.get_height()
    x, y = p.get_xy()
    if height > 0:
        ax.text(x + width/2,
                y + height/2,
                f'{height:.1f}%',
                horizontalalignment='center',
                verticalalignment='center')

plt.title('Churn by SeniorCitizen (% of Total)', fontsize=10)
plt.ylabel('Percentage of Total Population')
plt.xticks(rotation=0)
plt.show()
```

comparatively a greated percentage of people in senior citizen category have churned

```python
import matplotlib.pyplot as plt
import seaborn as sns

cols = ['PhoneService', 'MultipleLines', 'InternetService',
        'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']

fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 15))
axes = axes.flatten()

for i, col in enumerate(cols):

    sns.countplot(x=col, data=df, hue='Churn', ax=axes[i], palette='viridis')


    axes[i].set_title(f'Churn by {col}', fontsize=12)
    axes[i].set_xlabel('') # Clear x-label to keep it clean
    axes[i].legend(title='Churn', loc='upper right')

plt.tight_layout()
plt.show()
```

The analysis reveals that OnlineSecurity and TechSupport are the strongest retention drivers, as customers with these services are significantly more loyal. Conversely, Fiber Optic users represent the highest churn risk, suggesting that high-speed tiers may face issues with pricing or service stability. Standard utilities like PhoneService and Streaming have a negligible impact on retention, acting as expected features rather than factors that prevent customer turnover.

```python

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df.dropna(subset=['TotalCharges'], inplace=True)

df['Churn_Numeric'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

for the further analysis,we must ensure our data types are correct, specifically for TotalCharges, which often contains empty strings.

```python
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.show()
```

mathematical relationship between numerical variables and churn.

```python
# List of service columns
services = ['PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
            'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

df['Total_Services'] = (df[services] == 'Yes').sum(axis=1)

plt.figure(figsize=(10, 5))
sns.barplot(x='Total_Services', y='Churn_Numeric', data=df, palette='viridis')
plt.title('Churn Rate vs. Number of Services Subscribed')
plt.ylabel('Churn Probability')
plt.show()
```

it seems like as "Total Services" increases, churn probability decreases.

```python
def tenure_cohort(t):
    if t <= 12: return '0-1 Year'
    elif t <= 24: return '1-2 Years'
    elif t <= 48: return '2-4 Years'
    else: return 'Over 4 Years'

df['Tenure_Group'] = df['tenure'].apply(tenure_cohort)

plt.figure(figsize=(10, 5))
sns.countplot(x='Tenure_Group', hue='Churn', data=df, order=['0-1 Year', '1-2 Years', '2-4 Years', 'Over 4 Years'])
plt.title('Churn Distribution by Tenure Cohort')
plt.show()
```

Raw tenure is hard to read; buckets tell a story about customer loyalty stages(Cohort Analysis)

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title('Monthly Charges Outliers')

plt.subplot(1, 2, 2)
sns.boxplot(x='Churn', y='TotalCharges', data=df)
plt.title('Total Charges Outliers')
plt.show()
```

This shows  care about data integrity and understand the financial distribution

```python
df['CLV'] = df['MonthlyCharges'] * df['tenure']

plt.figure(figsize=(10, 5))
sns.kdeplot(df[df['Churn'] == 'Yes']['CLV'], label='Churned', fill=True)
sns.kdeplot(df[df['Churn'] == 'No']['CLV'], label='Retained', fill=True)
plt.title('Customer Lifetime Value Distribution: Churned vs Retained')
plt.legend()
plt.show()
```
